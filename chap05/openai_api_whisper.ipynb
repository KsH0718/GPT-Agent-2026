```python
from openai import OpenAI
from dotenv import load_dotenv  
import os
import textwrap

# .env 파일의 설정을 환경변수로 불러오기
load_dotenv()

# 환경변수에서 API 키 읽기
api_key = os.getenv("OPENAI_API_KEY")

if not api_key:
    raise ValueError("OPENAI_API_KEY 환경 변수가 설정되지 않았습니다. .env 파일을 확인하세요.")

# OpenAI API 요청을 보낼 객체 만들기
client = OpenAI(api_key=api_key)
```

```python
audio_file_path = "./audio/lsy_audio_2023_58s.mp3"

if not os.path.isfile(audio_file_path):
    raise FileNotFoundError(
        f"음성 파일을 찾지 못했습니다: {audio_file_path}\n"
        f"현재 작업 폴더: {os.getcwd()}"
    )

# 음성 파일을 바이너리 읽기 모드로 열기 
with open(audio_file_path, "rb") as audio_file:
    transcription = client.audio.transcriptions.create(
        file=audio_file,
        model="whisper-1"
    )

# 변환 결과에서 텍스트 가져오기
text = transcription.text

# 화면에서 읽기 편하도록 줄바꿈하기 
formatted_text = "\n\n".join(
    textwrap.fill(paragraph, width=60)
    for paragraph in text.split("\n\n")
)

print(formatted_text)
```
